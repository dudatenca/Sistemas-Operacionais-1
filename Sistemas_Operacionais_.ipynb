# **Aula 8 — Gerenciamento de Memória na Prática**

# Atividade 11 - Simulador de Gerenciamento de Memória.

```python


def fifo(paginas, frames):
    memoria = []
    faltas = 0
    hits = 0

    print("\n--- FIFO ---")

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"
        else:
            faltas += 1
            resultado = "PAGE FAULT"

            if len(memoria) >= frames:
                memoria.pop(0)

            memoria.append(pagina)

        print(f"Página {pagina} -> {memoria} | {resultado}")

    return faltas, hits


def lru(paginas, frames):
    memoria = []
    faltas = 0
    hits = 0

    print("\n--- LRU ---")

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"

            memoria.remove(pagina)
            memoria.append(pagina)
        else:
            faltas += 1
            resultado = "PAGE FAULT"

            if len(memoria) >= frames:
                memoria.pop(0)

            memoria.append(pagina)

        print(f"Página {pagina} -> {memoria} | {resultado}")

    return faltas, hits


# Entrada de dados
entrada = input("Digite as páginas separadas por espaço: ")
paginas = [int(valor) for valor in entrada.split()]

frames = int(input("Digite a quantidade de frames: "))

if frames <= 0 or len(paginas) == 0:
    print("Digite uma quantidade válida de frames e páginas.")
else:
    # Executando os algoritmos
    faltas_fifo, hits_fifo = fifo(paginas, frames)
    faltas_lru, hits_lru = lru(paginas, frames)

    total = len(paginas)

    # Resultados
    print("\n--- RESULTADOS ---")

    print("\nFIFO")
    print("Page Faults:", faltas_fifo)
    print("Hits:", hits_fifo)
    print(f"Taxa de Page Faults: {faltas_fifo / total * 100:.2f}%")
    print(f"Taxa de Hits: {hits_fifo / total * 100:.2f}%")

    print("\nLRU")
    print("Page Faults:", faltas_lru)
    print("Hits:", hits_lru)
    print(f"Taxa de Page Faults: {faltas_lru / total * 100:.2f}%")
    print(f"Taxa de Hits: {hits_lru / total * 100:.2f}%")

    # Comparação
    print("\n--- COMPARAÇÃO ---")

    if faltas_fifo < faltas_lru:
        print("FIFO apresentou menos Page Faults.")
    elif faltas_lru < faltas_fifo:
        print("LRU apresentou menos Page Faults.")
    else:
        print("FIFO e LRU tiveram o mesmo resultado.")
```

# Atividade 13 - Questões para análise

**1. Qual é a diferença entre um HIT e um PAGE FAULT?**
Um HIT acontece quando a página solicitada já está na memória. Já o PAGE FAULT acontece quando a página não está na memória e precisa ser carregada.

**2. Qual é a principal diferença entre FIFO e LRU?**
O FIFO remove a página que entrou primeiro na memória, enquanto o LRU remove a página que está há mais tempo sem ser utilizada.

**3. Um page fault significa necessariamente que existe um erro no programa? Explique.**
Não. O page fault é uma situação normal que acontece quando o programa precisa acessar uma página que ainda não está carregada na memória RAM.

**4. Aumentar a quantidade de frames sempre reduz o número de page faults no FIFO? Utilize o experimento da Anomalia de Belady para justificar.**
Não. Na Anomalia de Belady, podemos observar que o FIFO apresentou 9 page faults com 3 frames e 10 page faults com 4 frames. Isso mostra que aumentar os frames nem sempre melhora o resultado.

**5. Em quais situações FIFO e LRU podem produzir resultados diferentes?**
Quando uma página é acessada novamente, o LRU atualiza sua ordem de utilização, enquanto o FIFO mantém a ordem de entrada. Por isso, eles podem escolher páginas diferentes para remover.

**6. Por que o algoritmo Ótimo normalmente apresenta um bom resultado?**
Porque ele escolhe para remover a página que vai demorar mais para ser utilizada novamente, evitando substituições desnecessárias.

**7. Por que um sistema operacional real não consegue utilizar perfeitamente o algoritmo Ótimo?**
Porque ele precisaria saber quais páginas serão acessadas no futuro, e o sistema operacional não consegue prever isso com certeza.

**8. Execute pelo menos três sequências diferentes e registre qual algoritmo apresentou menos page faults em cada uma.**
Fiz três testes utilizando 3 frames:
- Teste 1: sequência [1, 2, 3, 1, 4, 2, 5, 1]. O FIFO apresentou 6 page faults e o LRU apresentou 7. Nesse caso, o FIFO foi melhor.
- Teste 2: sequência [7, 0, 1, 2, 0, 3, 0, 4, 2, 3, 0, 3, 2]. O FIFO apresentou 10 page faults e o LRU apresentou 9. Nesse caso, o LRU foi melhor.
- Teste 3: sequência [1, 2, 3, 4, 1, 2, 5, 1, 2, 3, 4, 5]. O FIFO apresentou 9 page faults e o LRU apresentou 10. Nesse caso, o FIFO foi melhor.
Com esses testes, percebi que nenhum dos dois algoritmos é sempre melhor. O resultado depende da sequência de páginas utilizada.
